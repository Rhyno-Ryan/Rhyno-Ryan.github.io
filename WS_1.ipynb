# 2. Cleaning Data

When dealing with sets of data, whether it come from a survey you took, or from an online resource it may not have formatting you would like. Or, it may have information you do not care about. Our goal in this note book will be to touch on the basics of cleaning data

- Removing duplicates
- Removing NaN's (Filling them in with a blank string)
- Removing columns

The basic commands will be



> df_foo.drop_duplicates()

> df_foo.fillna("")

> df_foo.drop(columns = "COLUMN_NAME")



## Example 2.1 -- Caller data

In this example we will examine sample caller data (See [1] in Resources). What are things we notice about the data below?

````python
df_call_list
````

Notice that this set of data has duplicates, NaN's, a useless column, and inconsistent formatting. For the sake of time, we will only worry about duplicates, NaN's, and the useless column. However [1] covers how to clean all of these cases.

We will first
1. Remove Duplicates
2. Repalces NaN's with empty strings
3. Remove Columns we do not need


````python
# Saving the unclean call list to compare later
df_call_list_old = df_call_list

# Removing Duplicates
df_call_list = df_call_list.drop_duplicates()

df_call_list
````

````python
# Removing NaN's (Replacing the NaN's)
df_call_list = df_call_list.fillna("")

#Visualize call list
df_call_list
````

````python
# Removing irrelevant columns
df_call_list = df_call_list.drop(columns= ["Not_Useful_Column","Last_Name","Phone_Number", "Address"])

#Visualize call list
df_call_list
````

As an extra clean up for the next section, we can remove rows with empty answers, or "N/a" as an answer. Then, standardize the answers.

````python
# Make all the Yes -> Y
#              No -> N
# in "Paying Customer" Column
df_call_list["Paying Customer"] = df_call_list["Paying Customer"].str.replace('Yes',"Y")
df_call_list["Paying Customer"] = df_call_list["Paying Customer"].str.replace('No',"N")

# Make all the Yes -> Y
#              No -> N
# in "Do_Not_Contact" Column
df_call_list["Do_Not_Contact"] = df_call_list["Do_Not_Contact"].str.replace('Yes',"Y")
df_call_list["Do_Not_Contact"] = df_call_list["Do_Not_Contact"].str.replace('No',"N")

#Visualize call list
df_call_list
````

````python
#Filtering through the call_list and column "Paying customer" to check for an empty string
for x in df_call_list.index:
  if (df_call_list.loc[x, "Paying Customer"] == '') or (df_call_list.loc[x, "Do_Not_Contact"] == ''):
    df_call_list.drop(x, inplace=True)

#Filtering through the call_list and column "Paying customer" to check for an "N/a"
for x in df_call_list.index:
  if df_call_list.loc[x, "Paying Customer"] == 'N/a' or df_call_list.loc[x, "Do_Not_Contact"] == 'N/a':
    df_call_list.drop(x, inplace=True)

#Visualize call list
df_call_list
````

## Example 3.2 -- Time Series

In a time series, we want to visualize a measured value over time. In this partiuclar example we will visualize Microsoft stock close value every week.

> px.line(df_foo, x='Date', y='plot_vals')



````python
df_MSFT_plot = df_MSFT.reset_index()
# Flatten multi-level columns from yfinance output
df_MSFT_plot.columns = ['_'.join(col).strip() if isinstance(col, tuple) and col[1] else col[0] for col in df_MSFT_plot.columns.values]

#Notice this is tuple
print(df_MSFT.columns)

#Notice this is an arrary of names
print(df_MSFT_plot.columns)

#Creating time series line plot
fig_time_ser = px.line(df_MSFT_plot, x='Date', y='Close_MSFT')

#Visualizing time series chart
fig_time_ser
````
